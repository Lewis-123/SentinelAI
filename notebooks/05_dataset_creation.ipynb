```python
from machine_learning.data_pipeline.dataset_builder import DatasetBuilder
from machine_learning.data_pipeline.risk_label import calculate_risk


builder = DatasetBuilder()


weather = builder.load_weather(
    "../data/raw/weather/weather_sample.csv"
)


population = builder.load_population(
    "../data/raw/population/population.csv"
)


dataset = builder.merge_data(
    weather,
    population
)


dataset["risk_level"] = dataset.apply(
    calculate_risk,
    axis=1
)


dataset
```
