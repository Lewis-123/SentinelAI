```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

dataset_path = "../data/processed/risk_training_dataset.csv"

df = pd.read_csv(dataset_path)
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="risk_level",
    y="rainfall"
)

plt.title(
    "Rainfall Distribution by Risk Level"
)

plt.xlabel(
    "Risk Level"
)

plt.ylabel(
    "Rainfall (mm)"
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="risk_level",
    y="temperature"
)

plt.title(
    "Temperature Distribution by Risk Level"
)

plt.xlabel(
    "Risk Level"
)

plt.ylabel(
    "Temperature (°C)"
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="risk_level",
    y="poverty_rate"
)

plt.title(
    "Poverty Rate Across Risk Categories"
)

plt.xlabel(
    "Risk Level"
)

plt.ylabel(
    "Poverty Rate (%)"
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="risk_level",
    y="population"
)

plt.title(
    "Population Exposure by Risk Level"
)

plt.xlabel(
    "Risk Level"
)

plt.ylabel(
    "Population"
)

plt.show()
```

```python
numeric_features = df.select_dtypes(
    include=np.number
)


plt.figure(figsize=(10,6))

sns.heatmap(
    numeric_features.corr(),
    annot=True
)

plt.title(
    "Feature Correlation Matrix"
)

plt.show()
```

```python
risk_mapping = {
    "LOW": 0,
    "MEDIUM": 1,
    "HIGH": 2
}


df["risk_score"] = df["risk_level"].map(
    risk_mapping
)


df.head()
```

```python
df.to_csv(
    "../data/processed/risk_training_dataset_encoded.csv",
    index=False
)
```
